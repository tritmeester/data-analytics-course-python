Oefening 1.3: Voeg kolommen toe aan movies_df voor de profit van de film (i.e. revenue - budget) en voor het jaar waarin de film uit kwam (i.e. alleen het jaar, niet de volledige datum).

```python
enriched_movies_df = movies_df.assign(profit = pd.col('revenue') - pd.col('budget'),
                                      release_year = pd.col('release_date').dt.year) # n.b.: Requires release_date dtype being fixed
enriched_movies_df
```

Oefening 1.4: Gebruik de .rank functie (i.e. niet .sort_values/.nlargest), om de top 10 winstgevende films te vinden die gereleased zijn 2015. Herken je ze?

```python
(
    enriched_movies_df[['movie_id', 'title', 'release_year', 'profit']]
    .loc[pd.col('release_year') == 2015]
    .assign(profit_rank = pd.col('profit').rank(ascending = False))
    .loc[pd.col('profit_rank') <= 10]
)
```

Oefening 2.1: Bereken de gemiddelde speelduur per jaar voor de laatste 25 jaar. Wat valt je op?

```python
runtime_by_year = (
    movies_df
    .assign(release_year = pd.col('release_date').dt.year)
    .groupby("release_year", as_index = False)
    .agg(average_runtime = ('runtime', 'mean'))
    .sort_values(by = "release_year", ascending = False)
)

runtime_by_year.head(25)
```

Oefening 2.2: Neem alleen films mee waar minstens 100 keer op gestemd is. Geef voor elk jaar antwoord op de volgende vraag: Hebben films met een hoger dan gemiddelde speelduur (binnen dat jaar!) een hogere gemiddelde vote_average dan films met een minder dan gemiddelde speelduur? Hint: Gebruik window functions.

```python
(movies_df
    .loc[pd.col('vote_count') >= 100]
    .assign(
        release_year = pd.col('release_date').dt.year,
        average_runtime_in_release_year = lambda df:  df.groupby('release_year')['runtime'].transform('mean'),
        runtime_longer_than_average_in_year = pd.col('runtime') > pd.col('average_runtime_in_release_year')
    )    
    .groupby(
        by = ['release_year', 'runtime_longer_than_average_in_year'],
        as_index = False
    ).agg(
        vote_average = ('vote_average', 'mean')
    ).pivot(index = ['release_year'], columns=['runtime_longer_than_average_in_year'], values='vote_average')
    .reset_index().rename(columns = {False: 'average_vote_shorter_than_average_runtime', True: 'average_vote_longer_than_average_runtime'})
    .sort_values('release_year', ascending = False)
)
```

Oefening 2.3: In welke genres wordt gemiddeld de meeste verschillende talen gesproken? Hiervoor heb je het '../data/movies_spoken_languages.csv' bestand nodig.

```python
spoken_languages_df = pd.read_csv('../../data/movies_spoken_languages.csv') # Adjust the path to the folder you're running this from
```

```python
spoken_languages_df.sample(10)
```

```python
print(spoken_languages_df.dtypes)
```

```python
number_of_languages_df = spoken_languages_df.groupby('movie_id').agg(number_of_languages = ('language_id', 'count'))
number_of_languages_df
```

```python
movies_df['genres']
```

```python
(movies_df
    .explode('genres') # n.b.: Requires genres dtype being fixed <- if not yet fixed will take extra effort
    .rename(columns = {'genres': 'genre'})
    .dropna(subset = ['genre']) # Pandas inserts nulls when exploding empty lists
    .merge(number_of_languages_df, on = 'movie_id')
    .groupby('genre')
    .agg(average_number_of_languages = ("number_of_languages", "mean"))
    .sort_values('average_number_of_languages', ascending = False)
)
```
